## 3. Data Preparation

The `Date` column is converted from text into a datetime format so that the dataset can be analyzed by month and quarter.

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Date"].dtype
```

## 4. Monthly Sales Trend

Monthly sales are analyzed to identify periods of high and low sales activity throughout the available dataset.

```python
monthly_sales = df.groupby(
    df["Date"].dt.to_period("M")
)["Total Amount"].sum()

monthly_sales
```

```python
plt.figure(figsize=(12, 6))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

### Monthly Sales Observation

Monthly sales fluctuate considerably throughout the dataset. Sales reached their highest level in **May 2023**, with a total sales amount of **53,150**.

Sales then declined in the following months, reaching a low of **23,620** in September 2023 before recovering to **46,580** in October.

The dataset also contains January 2024, where sales dropped sharply to **1,530**. This appears to represent only a small number of transactions in the available data rather than necessarily indicating a long-term decline in business performance.

## 5. Quarterly Sales Trend

Quarterly sales are analyzed to identify broader changes in sales performance across the year.

```python
plt.figure(figsize=(10, 6))

quarterly_sales.plot(kind="line", marker="o")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

### Quarterly Sales Observation

Quarterly sales also show noticeable variation throughout the dataset. Sales increased from **108,500 in Q1 2023** to **123,735 in Q2 2023**, before declining to **96,045 in Q3 2023**.

Sales then recovered and reached the highest quarterly total of **126,190 in Q4 2023**.

The available Q1 2024 data contains only **1,530** in sales, which is substantially lower than the previous quarters. Since the dataset only includes January 2024, this figure should not be interpreted as a complete-quarter performance measure.

## 6. Customer Demographics

Customer demographics are examined using age groups and gender to understand how transactions are distributed across different customer segments.

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[17, 24, 34, 44, 54, 64],
    labels=["18–24", "25–34", "35–44", "45–54", "55–64"]
)

age_groups = df["Age Group"].value_counts().sort_index()

age_groups
```

```python
plt.figure(figsize=(10, 6))

age_groups.plot(kind="bar")

plt.title("Transaction Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Age Group Observation

The transaction distribution is relatively balanced across the age groups, although some groups have slightly more transactions than others.

The **45–54 age group** has the highest number of transactions at **225**, followed by **55–64 with 216** and **35–44 with 207**.

The **18–24 age group** has the fewest transactions at **149**.

Overall, the data suggests that transactions are distributed across a broad range of adult age groups rather than being concentrated in only one age category.

```python
gender_counts = df["Gender"].value_counts()

gender_counts
```

```python
plt.figure(figsize=(8, 5))

gender_counts.plot(kind="bar")

plt.title("Transaction Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Gender Distribution Observation

The gender distribution is almost evenly balanced in the dataset. **Female transactions account for 510 transactions**, while **male transactions account for 490 transactions**.

This indicates that neither gender dominates the transaction data significantly.

Therefore, gender-based differences in purchasing behavior may require further analysis of metrics such as spending and product preferences rather than transaction counts alone.

## 7. Product Category Analysis

Product categories are analyzed using both revenue and quantity sold to determine which categories contribute most to overall sales and transaction volume.

```python
category_revenue = df.groupby(
    "Product Category"
)["Total Amount"].sum().sort_values(ascending=False)

category_revenue
```

```python
plt.figure(figsize=(9, 6))

category_revenue.plot(kind="bar")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Product Category Revenue Observation

Electronics generated the highest total sales amount at **156,905**, followed closely by Clothing at **155,580**. Beauty generated the lowest total sales amount at **143,515**.

The difference between Electronics and Clothing is relatively small, while Beauty records a noticeably lower total.

This suggests that Electronics and Clothing are the strongest revenue-generating categories in the available dataset.

```python
category_quantity = df.groupby(
    "Product Category"
)["Quantity"].sum().sort_values(ascending=False)

category_quantity
```

```python
plt.figure(figsize=(9, 6))

category_quantity.plot(kind="bar")

plt.title("Units Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Units Sold Observation

Clothing recorded the highest quantity sold, with **894 units**, followed by Electronics with **849 units** and Beauty with **771 units**.

Comparing quantity sold with revenue reveals an important difference: **Clothing sold the most units, but Electronics generated the highest total revenue**.

This suggests that Electronics products may generate more revenue per unit than Clothing products.

Because the dataset does not contain individual product names, the analysis is limited to product-category performance rather than a true Top 10 product ranking.

### Dataset Limitation: Top 10 Products

The internship checklist includes identifying the top 10 best-selling products. However, this dataset contains a `Product Category` column but does not contain an individual product-name column.

## 8. Correlation Analysis

Correlation analysis is used to examine the relationships between the main numerical business variables: age, quantity, price per unit, and total transaction amount.

The transaction ID is excluded because it is an identifier rather than a meaningful business measurement.

```python
numeric_data = df[
    ["Age", "Quantity", "Price per Unit", "Total Amount"]
]

correlation_matrix = numeric_data.corr()

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()

plt.show()
```

### Correlation Heatmap Observation

The strongest relationship in the numerical variables is between **Price per Unit and Total Amount**, with a correlation of **0.852**. This strong positive relationship indicates that higher-priced items are generally associated with higher transaction amounts.

**Quantity and Total Amount** have a moderate positive correlation of **0.374**, suggesting that transactions involving more items tend to have higher total amounts, although the relationship is not as strong as the relationship between price and total amount.

Age has very weak correlations with Total Amount (**-0.061**), Quantity (**-0.024**), and Price per Unit (**-0.038**). This suggests that age has little linear relationship with these purchasing variables in this dataset.

## 9. Additional Analysis: Average Transaction Value by Gender

An additional analysis was performed to determine whether male and female customers have noticeably different average transaction values.

```python
avg_transaction_by_gender = df.groupby(
    "Gender"
)["Total Amount"].mean().sort_values(ascending=False)

avg_transaction_by_gender
```

```python
plt.figure(figsize=(8, 5))

avg_transaction_by_gender.plot(kind="bar")

plt.title("Average Transaction Value by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Transaction Value")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

### Average Transaction Value Observation

The average transaction value is almost identical for female and male customers.

Female customers have an average transaction value of approximately **456.55**, while male customers have an average of approximately **455.43**.

The difference of about **1.12** is very small, suggesting that gender does not have a meaningful effect on average transaction value in this dataset.

This indicates that marketing and sales strategies should not rely heavily on gender-based assumptions when targeting customers. Other factors, such as product category, price, and purchasing quantity, appear more useful for understanding transaction value.

## 10. Conclusion and Business Recommendations

### Conclusion

The exploratory data analysis revealed several important patterns in the retail sales dataset. Sales varied considerably across months and quarters, with **May 2023** recording the highest monthly sales and **Q4 2023** recording the highest quarterly sales.

Among the product categories, **Electronics generated the highest revenue at 156,905**, while **Clothing recorded the highest quantity sold at 894 units**. This shows that the category selling the most units is not necessarily the category generating the most revenue.

The correlation analysis showed a strong positive relationship between **Price per Unit and Total Amount (0.852)**, while Quantity had a moderate positive relationship with Total Amount (**0.374**). Customer age showed very weak relationships with transaction value and quantity.

In addition, average transaction values for female and male customers were almost identical, suggesting that gender is not a major factor affecting transaction value in this dataset.

### Business Recommendations

1. **Prioritize high-value Electronics products.**  
   Electronics generated the highest total revenue, so the business should maintain strong inventory levels for this category and consider promoting higher-value Electronics products to increase revenue.

2. **Use seasonal sales patterns for inventory and promotions.**  
   Sales varied significantly across the year, with strong performance in May and Q4 2023. The business should prepare sufficient inventory and targeted promotional campaigns ahead of historically stronger periods while investigating the causes of weaker periods such as September.

3. **Focus on pricing and product value.**  
   Price per Unit had the strongest relationship with Total Amount. The business should analyze pricing carefully and identify products where customers are willing to purchase at higher price points while maintaining competitive pricing.

4. **Avoid relying heavily on gender-based targeting.**  
   Average transaction values for male and female customers were almost identical. Marketing decisions should therefore focus more on purchasing behavior, product category, and transaction value rather than assuming major differences based on gender.

5. **Encourage larger purchases.**  
   Quantity had a moderate positive relationship with Total Amount. The business could experiment with product bundles, cross-selling, and quantity-based promotions to encourage customers to purchase more items per transaction.

### Dataset Limitations

The dataset does not contain individual product names, so a genuine Top 10 best-selling product analysis could not be performed without inventing product-level information.

The dataset also contains only part of January 2024. Therefore, the January 2024 and Q1 2024 figures should not be interpreted as complete-period performance.
